```python
import numpy as n
import matplotlib.pyplot as plt
```

# Fix SNR ratio (just changing magnitude)

seeing what exposure time is required

```python
m1 = 10

m2 = n.array([10, 12, 16, 17, 18, 19])

F_ratio = 10**( (m2 - m1)/-2.5) #F2/F1

t1 = 120 #seconds

t2 = n.sqrt((F_ratio**-1) * t1)
```

```python
plt.figure('x')
plt.plot(m2, t2, 'ko')
plt.ylabel('Exposure Time (seconds)')
plt.xlabel('Magnitude')
plt.show()
```

# SNR, now adjusting for aperture (2.3 m - Bok, vs 3.1 m KAST)

```python
D1 = 3.1 #meters
D2 = 2.3 #meters

t2_d2 = ((D1/D2)**2) * t2
```

```python
plt.figure('Scaled to New Telescope')
plt.plot(m2, t2, 'ko', label = 'KAST')
plt.plot(m2, t2_d2, 'ro', label = "BOK")
plt.ylabel('Exposure Time (seconds)')
plt.xlabel('Magnitude')
plt.legend()
plt.show()
```
