```python
import matplotlib.pyplot as plt
import numpy as np
%matplotlib notebook 
# the above line may not work for all of you
```

```python
#From Taylor book chapter 5. We have 10 observations of a length of something
lengths = np.array([26,24,26,28,23,24,25,24,26,25])
```

```python
lengths_std = np.std(lengths) #calculating the standard deviation
lengths_mean = np.mean(lengths) #calculating the mean

print('standard deviation: ' + str(lengths_std))
print('mean: ' + str(lengths_mean))
```

```python
labelsize = 20 # size of my labels on my plots
ticksize = 14  # the size of the numerical ticks of my plots

plt.figure('histogram of lengths counts')
#plt.hist(lengths, density=True)
plt.hist(lengths, density=False)
plt.xlabel('Length (cm)', size = labelsize)
plt.ylabel('Frequency \n (# of observations)', size = labelsize)
plt.xticks(size = ticksize)
plt.yticks(size = ticksize)
plt.xlim([22,28])
plt.tight_layout()
plt.show()
```
