# Choosing an Observatory Site for Observations

List of observatory data avaliable in astropy here: https://github.com/astropy/astropy-data/blob/gh-pages/coordinates/sites.json Astropy is the module where astroplan is pulling some of its data.

We'll first establish a class that pulls the relevant information from the observatory where we want to observe at

```python
#getting coordinates, elevation, etc information for Mt. Bigelow
site = Observer.at_site("mtbigelow") # Mt. Bigelow
#site = Observer.at_site("kpno")      # Kitt Peak National Observatory
```

we'll print the site information along with the timezone. The default timezone is Coordinated Universal Time (UTC). The local time in Tucson is always 7 hours behind UT time.

```python
site
```

# Defining Targets in Astropy

In order to understand if a target is visible we first need to create a class for that target by inputting the target's right acension (longitude on the celestial sphere) and declination (latitude on the celestial sphere)

```python
from astropy.coordinates import SkyCoord
import astropy.units as u

target_name = 'HIP99770'
target_ra = "8h13m51.8490476736s"  #hours, minutes, seconds
target_dec = "63d10m18.835641012s"  #degrees, minutes, seconds

#creating the class object with our target's information
target = FixedTarget(coord=SkyCoord(ra=target_ra, dec=target_dec), name=target_name)
```

# Defining a Time Range for Observations

Midnight in Tucson time is 7:00 UTC so we are setting that as the middle point for where we want to define our time array. We can change the year, month, and day to plan our observations for the semestery. We are creating a linear array to cover up to 6 hours before midnight (sunset) and 6 hours after (sunrise)

```python
from astropy.time import Time
import numpy as np

#Time('YYYY-mm-dd')
obs_time_range = Time('2026-01-01 7:00:00') + np.linspace(-6, 7, 55)*u.hour 
```

# Plotting Visbility - Airmass and Altitude for a Target

Next we are plotting airmass and alititude vs UTC time below. The shaded area tells you how strong twilight is. An object is visible when the airmass or altitude is high enough for the telescope to view. Most telescopes cannot point directly at horizon or may have obsticles near the horizon that limit visibility. Some telescopes like the LBT cannot point above a certain airmass. You still need to check when the moon's bright/dark time is in the almanac.

We'll make a visibility plot for our single target for over a night

```python
from astroplan.plots import plot_airmass
import matplotlib.pyplot as plt

plot_airmass(target, site, obs_time_range, altitude_yaxis=True, brightness_shading=True)
```

# Plotting Visbility over Different Times

now we are going to plot three different time ranges side by side using subplots in matplotlib. We are plotting a night in January, February, and March. Notice how this target rises earlier and sets earlier each month over this time period. Plug in your preferred targets to understand which weekends or months are bad for your project.

```python
#Time('YYYY-mm-dd')
obs_time_range1 = Time('2026-01-01 7:00:00') + np.linspace(-6, 7, 55)*u.hour 
obs_time_range2 = Time('2026-02-01 7:00:00') + np.linspace(-6, 7, 55)*u.hour 
obs_time_range3 = Time('2026-03-01 7:00:00') + np.linspace(-6, 7, 55)*u.hour 
```

```python
plt.figure(figsize = (15, 5))
plt.subplot(131)
plot_airmass(target, site, obs_time_range1, altitude_yaxis=True)
plt.subplot(132)
plot_airmass(target, site, obs_time_range2, altitude_yaxis=True)
plt.subplot(133)
plot_airmass(target, site, obs_time_range3, altitude_yaxis=True)
plt.tight_layout()
```
