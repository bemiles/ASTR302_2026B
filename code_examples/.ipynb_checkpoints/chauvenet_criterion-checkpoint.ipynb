```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.integrate import quad

#Section 6.2 in observational astronomy book
```

```python
#gaussian function
def gaussian(x):
    mu = 1
    sig = 1
    return np.exp(-0.5 * ((x - mu) / sigma) ** 2) / (sigma * np.sqrt(2 * np.pi))
```

# Testing How Likely a Measurement Resides Beyond 2 sigma

```python
measurements = np.array([3.8, 3.5, 3.9, 3.9, 3.4, 1.8])
n_measurements = len(measurements) #number of measurements
```

```python
labelsize = 20 # size of my labels on my plots
ticksize = 14  # the size of the numerical ticks of my plots

plt.figure('histogram of lengths counts')
#plt.hist(lengths, density=True)
plt.hist(measurements, density=False, align = 'mid')
plt.xlabel('Length (cm)', size = labelsize)
plt.ylabel('Frequency \n (# of observations)', size = labelsize)
plt.xticks(size = ticksize)
plt.yticks(size = ticksize)
plt.tight_layout()
plt.show()
```

```python
mean = np.mean(measurements)
mean
```

```python
sigma = np.std(measurements)
sigma
```

```python
sigma_limit = 2
prob_in2sigma = 0.95 #probability within 2 sigma of the mean

prob_out2sigma = 1 - prob_in2sigma
prob_out2sigma
```

```python
expected_out2sigma = n_measurements * prob_out2sigma
expected_out2sigma
```

# Testing How Likely a Suspect Measurement Resides at its Calculated Sigma
if the variable expected_out is less than 0.5, you can reject the data point

```python
#5th index is the implied outlier it is much lower than the mean
sigma_sus = abs(measurements[5] - mean)/sigma
```

```python
sigma_sus
```

```python
#you need to integrate over a region to understand the probability of that measurment hapepening
prob_in, err = quad(gaussian, -sigma_sus, sigma_sus)
prob_in
```

```python
prob_out = 1 - prob_in
prob_out
```

```python
expected_out = n_measurements * prob_out
expected_out
```
