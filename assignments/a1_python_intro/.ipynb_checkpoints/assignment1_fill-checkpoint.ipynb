 # Creating and Plotting Probability Distributions in Python

We are going to use numpy's random number generator numpy.random to simulate different probability distributions. First we will work with the normal distribution and then the poisson distribution later.

https://numpy.org/doc/stable/reference/random/generated/numpy.random.normal.html

## Gaussian Distribution

```python
mu_normal = 10
sigma_normal = 5
sample_size = 10000

sample_normal = np.random.normal(mu_normal,sigma_normal, sample_size)
```

```python
print(sample_normal)
```

Now we can use the matplotlib package to plot arrays.  First import it, and configure it to draw within the python notebook

```python
import matplotlib.pyplot as plt
```

```python
n_bins = 100 #you can play with how many bins the plot has
plt.figure('Gaussian Distribution', figsize=(10,4))
plt.hist(sample_normal, align = 'mid', bins = n_bins )
plt.xlabel('Value')
plt.ylabel('Frequency (N)')
plt.title('Gaussian Distribution')
plt.show()
```

## Poisson Distribution

The Poisson distribution gives you the probability of encountering a certain number of events in a given period, if those events are occuring at a known constant mean rate.  This is the distribution that governs most 'counting' situations, including the collection of photons from astronomical sources.

The distribution has one parameter: the average number of events, often expressed as $\mu$:

$$ P(x) = \frac{\mu^\nu}{\nu!} e^{-\mu}$$

$\nu$ is not an input parameter, but the probability changes depending on how many events already occured

Use the Poisson distribution generator in numpy to generate a sample function with a mean rate of 10 for one hundred examples of the experiment.

```python
mu_poisson = 10
samples_poisson = np.random.poisson(mu_poisson, 100)
```

```python
print(samples_poisson)
```

```python
n_bins = 5 #you can play with how many bins the plot has
plt.figure(figsize=(10,4))
plt.hist(samples_poisson, bins = n_bins)
plt.xlabel('Number of Occurances')
plt.ylabel('Frequency (N)')
plt.title('Poisson Distribution')
plt.show()
```

# Extra: Reading in Astronomical Data and Information

The *astropy* package provides numerous astronomical utilities.  There are sub-packages within *astropy*, such as *constants* or *fits*, which contains useful astronomical constants or data handling functions.  You can import the subpackage like this:

```python
import astropy.constants
c = astropy.constants.c  #Retrieve the speed of light, and store in variable 'c'
print(c) #Print a variable using the print object
print(c.value) #Get just the value
```

now we will use astropy to read in a FITS (Flexible Image Transport System) file and plot an image. You can import a submodule by using from module import submodule

```python
from astropy.io import fits
```

now we'll call in data from a file. This is a picture of a star HIP 99770 from Shark NIR instrument on the Large Binocular Telescope. The star is behind a coronagraph and we were looking for a planet

```python
image_data = fits.getdata('hip99770.fits')
```

and print the dimensions of the file

```python
image_data.shape
```

this array is 2-dimensional because it is a picture (x,y) and each pixel has count/brightness value associated with it. We can also plot an image using matplotlib

```python
plt.figure('detector image')
plt.imshow(image_data)
plt.xlabel('x (pixel)')
plt.ylabel('y (pixel)')
plt.show()
```

try adjusting the vmin or vmax values in the cell below to see more structure in the image. Read the controls for matplotlib's imshow function: https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.imshow.html

```python
plt.figure('detector image adjusted')
plt.imshow(image_data, vmin = 10, vmax = 10000)
plt.xlabel('x (pixel)')
plt.ylabel('y (pixel)')
plt.show()
```
